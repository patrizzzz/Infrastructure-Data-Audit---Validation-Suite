## Relational SQL Queries & Visual Risk Profiling
This notebook queries the SQLite `infrastructure_audit.db` database to perform regional bridge condition risk analysis and SQL aggregations.

```python
import sys
import os
sys.path.insert(0, os.path.abspath('..'))

from src.db_manager import DatabaseManager

db_mgr = DatabaseManager('../data/processed/infrastructure_audit.db')

query = '''
SELECT 
    REGION,
    COUNT(BRIDGE_ID) AS Total_Bridges,
    SUM(CASE WHEN CONDITION = 'Poor' THEN 1 ELSE 0 END) AS Poor_Condition,
    SUM(CASE WHEN CONDITION = 'Bad' THEN 1 ELSE 0 END) AS Bad_Condition,
    ROUND(AVG(BR_LENGTH), 2) AS Avg_Length_Meters
FROM clean_bridges
GROUP BY REGION
ORDER BY Total_Bridges DESC
'''

regional_df = db_mgr.run_query(query)
regional_df
```
